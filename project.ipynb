```python
import pandas as pd
```

```python
df =pd.read_csv('huge_1M_titanic.csv')
```

```python
df.drop(columns=['PassengerId','Name','Age','Cabin','Ticket'],inplace=True)
```

```python
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)

df.info()
```

```python
df.isna().sum().sum()
```

```python
df
```

```python
df['Fare'] = df['Fare'].astype('int8')
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
df['Sex'] = le.fit_transform(df[['Sex']])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
df.isna().sum().sum()
```

```python
ohe = OneHotEncoder(sparse_output=False)
df['Embarked']=df['Embarked'].replace({'S':'Southampton','C':'Cherbourg','Q':'Queenstown'})
temp = pd.DataFrame(ohe.fit_transform(df[['Embarked']]),columns=ohe.get_feature_names_out())
```

```python
df.isna().sum().sum()
```

```python
df=pd.concat((df,temp),axis=1)
df.drop(columns='Embarked',inplace=True)
```

```python
df.info()
```

```python
df.head(5)
```

## Feature Encoding

```python
from sklearn.preprocessing import StandardScaler
```

```python
num_col = df.select_dtypes(include='number').columns
num_col = num_col.drop('Survived')

ss = StandardScaler()
df[num_col] = ss.fit_transform(df[num_col])
```

```python
with open('one_hot_encoder.pkl','wb') as file:
    pickle.dump(ohe,file)
with open('standard_scaler.pkl','wb') as file:
    pickle.dump(ss,file)
```

# Train Validation Test Split

```python
X = df.drop(columns='Survived')
```

```python
Y = df['Survived']
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_temp,Y_train,Y_temp = train_test_split(X,Y,test_size=0.36,random_state=42)

X_val,X_test,Y_val,Y_test = train_test_split(X_temp,Y_temp,test_size=0.5556,random_state=42)
```

```python
X_train.shape
```

```python
X_val.shape
```

```python
X_test.shape
```

```python
import tensorflow
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential([Dense(64,input_shape=(X_train.shape[1],),activation='relu'),
                    Dense(32,activation='relu'),
                    Dense(1,activation='sigmoid')]) #Last hidden layer
```

```python
opt = tensorflow.optimizers.Adam(learning_rate=0.01)
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    mode="min",
    restore_best_weights=True,
)
```

```python
model.fit(X_train,Y_train,validation_data=(X_val,Y_val),epochs=10,callbacks=[early_stopping])
```

```python
model.save('model.h5')
```

```python
model.predict(X_val, Y_val)
```
